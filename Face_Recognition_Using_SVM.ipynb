```python
import os
import face_recognition
```

```python
faces = []
labels = []
```

```python
datasets = '/content/Face_recognition'
for person in os.listdir(datasets):
  person_path = os.path.join(datasets,person)
  for image in os.listdir(person_path):
     image_path = os.path.join(person_path,image)
     img = face_recognition.load_image_file(image_path)
     encoding = face_recognition.face_encodings(img)
     if len(encoding) > 0:
        faces.append(encoding[0])
        labels.append(person)
print(faces[0].shape)
print(labels)
```

```python
from sklearn.svm import SVC
model = SVC(kernel='linear')
model.fit(faces,labels)
print("TRAINING COMPLETED !!!")
```

```python
import matplotlib.pyplot as plt

test_img = face_recognition.load_image_file('/content/Alia_Test_Image.jpg')
test_encoding = face_recognition.face_encodings(test_img)
print("encoding value: ", test_encoding)
if len(test_encoding) > 0:
  predict = model.predict(test_encoding)
  print(predict[0])
plt.figure(figsize=(5,5))
plt.imshow(test_img)
plt.axis("off")
plt.show()
```

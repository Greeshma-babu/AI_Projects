# Result

````python
# Install required packages
!pip install ultralytics moviepy -q


# Imports
from ultralytics import YOLO
from moviepy.editor import VideoFileClip
from google.colab.patches import cv2_imshow
import cv2


# Load Helmet Model
model = YOLO("helmet.pt")

# Check classes
print(model.names)


# Load Video
video_path = "Helment_IN_Work_Place.mp4"

clip = VideoFileClip(video_path)

print("Video Duration:", clip.duration, "seconds")


# Final counters
max_total = 0
max_with_helmet = 0
max_without_helmet = 0


# Process video frame by frame
for t in range(0, int(clip.duration)):

    print("Processing second:", t)

    # Read frame using MoviePy
    frame = clip.get_frame(t)

    # RGB to BGR
    frame = cv2.cvtColor(
        frame,
        cv2.COLOR_RGB2BGR
    )


    # YOLO Detection
    results = model(
        frame,
        imgsz=640,
        conf=0.25
    )


    with_helmet = 0
    without_helmet = 0


    # Count detections
    for box in results[0].boxes:

        cls = int(box.cls[0])

        if cls == 0:
            with_helmet += 1

        elif cls == 1:
            without_helmet += 1


    total = with_helmet + without_helmet


    # Store maximum detected count
    max_total = max(max_total, total)
    max_with_helmet = max(max_with_helmet, with_helmet)
    max_without_helmet = max(max_without_helmet, without_helmet)



    # Draw YOLO boxes
    annotated_frame = results[0].plot()


    # Add count display
    cv2.putText(
        annotated_frame,
        f"Total Persons: {total}",
        (20,40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,255,0),
        2
    )


    cv2.putText(
        annotated_frame,
        f"With Helmet: {with_helmet}",
        (20,80),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,255,0),
        2
    )


    cv2.putText(
        annotated_frame,
        f"Without Helmet: {without_helmet}",
        (20,120),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,0,255),
        2
    )


    # Display frame in Colab
    cv2_imshow(annotated_frame)



# Final Result

print("\n========== FINAL RESULT ==========")

print("Maximum Persons Detected :", max_total)

print("Persons With Helmet      :", max_with_helmet)

print("Persons Without Helmet   :", max_without_helmet)
````

```python
import nltk
```

# Tokenization

```python
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize, line_tokenize
text = """Hello All.
Let us learn the AI and ML course. Let us start learning the lesson faster."""

word_token = word_tokenize(text)
sent_tokenize = sent_tokenize(text)
line_tokenize = line_tokenize(text)
print("word_token:", word_token)
print("sent_tokenize:", sent_tokenize)
print("line_tokenize:", line_tokenize)
# NLTK does not provide a built-in subword tokenizer like BPE, WordPiece, or SentencePiece.
```

# Remove Stop Words

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words_result = []
stop_words = set(stopwords.words("english"))
for word in word_token :
  if word.lower() not in stop_words :
    stop_words_result.append(word)
print("After stop words removal = ",stop_words_result)
```

# Stemming

```python
from nltk.stem import PorterStemmer
stem = PorterStemmer()
stem_result = []
for word in stop_words_result:
  stem_result.append(stem.stem(word))
print("Stemmed = ",stem_result)
```

# Lemmization

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lem_result = []
for word in stop_words_result:
  lem_result.append(lemmatizer.lemmatize(word))
print("Lemmatized = ",lem_result)
```

# POS Tagging

```python
 nltk.download('averaged_perceptron_tagger_eng')
```

```python
pos_tagged = nltk.pos_tag(lem_result)
print(pos_tagged)
```

# NER

```python
nltk.download('words')
```

```python
print(nltk.ne_chunk(pos_tagged))
```

# Text Classification


```python
from nltk import FreqDist
text = "AI AI AI NLP AI NLP NLTK"
words = word_tokenize(text)
freq_dist = FreqDist(words)
print(freq_dist)
print(freq_dist.most_common())
```

```python
from nltk.classify import NaiveBayesClassifier
# NaiveBayier not expecting raw data - need trained data
train_data = [
    ({'word': 'AI'}, 'Artifical Intelligence'),
    ({'word': 'NLP'}, 'NL'),
    ({'word': 'NLTK'}, 'NL')
]
naivebayes = NaiveBayesClassifier.train(train_data)

# Test data
test_word = {'word': 'AI'}

# Prediction
result = naivebayes.classify(test_word)
print(result)
```

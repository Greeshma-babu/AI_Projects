```python
# Predict the data based on training data

# sklearn - Scikit learn  - CountVectorizer will convert the raw text into number based on word occurance.
from sklearn.feature_extraction.text import CountVectorizer

# Sklearn - svm (support vector machine) SVC - support vector classifier - Support Multiclass Classification
from sklearn.svm import SVC
```

```python
class Category:
    BOOK = "BOOK"
    SHOE = "SHOE"
    CLOTH = "CLOTH"
    PLACE = "PLACE"

x_train = [
    # BOOK
    "alchemist is a good book",
    "I enjoy reading novels",
    "this author writes amazing novel",
    "the library has many books",
    "I bought a new story book",
    "reading books improves knowledge",
    "the textbook is very useful",
    "she loves mystery novels",
    "I finished the book yesterday",
    "this book has an interesting plot",

    # SHOE
    "I love running shoes",
    "these sneakers are comfortable",
    "I bought new sports shoes",
    "my shoes are very stylish",
    "these boots are durable",
    "I wear shoes for jogging",
    "the shoe store has discounts",
    "these leather shoes look elegant",
    "my sneakers are brand new",
    "comfortable shoes help walking",

    # CLOTH
    "I bought a new shirt",
    "this dress looks beautiful",
    "the jacket keeps me warm",
    "I need a pair of jeans",
    "the cloth material is soft",
    "she wears a colorful saree",
    "this t shirt is comfortable",
    "I purchased a winter coat",
    "the fabric quality is excellent",
    "he likes fashionable clothes",

    # PLACE
    "Kerala is a beautiful place",
    "I visited the beach yesterday",
    "the park is very peaceful",
    "Paris is a famous city",
    "we went to a tourist destination",
    "the restaurant is near my house",
    "I traveled to a mountain resort",
    "the temple attracts many visitors",
    "this village is very scenic",
    "the museum is worth visiting"
]

y_train = [
    # BOOK
    Category.BOOK, Category.BOOK, Category.BOOK, Category.BOOK, Category.BOOK,
    Category.BOOK, Category.BOOK, Category.BOOK, Category.BOOK, Category.BOOK,

    # SHOE
    Category.SHOE, Category.SHOE, Category.SHOE, Category.SHOE, Category.SHOE,
    Category.SHOE, Category.SHOE, Category.SHOE, Category.SHOE, Category.SHOE,

    # CLOTH
    Category.CLOTH, Category.CLOTH, Category.CLOTH, Category.CLOTH, Category.CLOTH,
    Category.CLOTH, Category.CLOTH, Category.CLOTH, Category.CLOTH, Category.CLOTH,

    # PLACE
    Category.PLACE, Category.PLACE, Category.PLACE, Category.PLACE, Category.PLACE,
    Category.PLACE, Category.PLACE, Category.PLACE, Category.PLACE, Category.PLACE
]
```

```python
model = SVC(kernel='linear')
```

```python
vectorize = CountVectorizer(binary = True)
x_train_vectorized = vectorize.fit_transform(x_train)
model.fit(x_train_vectorized, y_train)
# We are vectorizing the y_train. Since, we model is lable the target values (y_train)
x_test = ["I bought a novel.","These sneakers are comfortable","I purchased a cotton shirt!", "We visited a beautiful waterfall. I loved it."]
predict = model.predict(vectorize.transform(x_test))
print(predict)
```

```python
# STEMMING IS THE PROCESS OF REMOVING WORD TO THEIR ROOT FORM BY REMOVNG SUFFIES AND PREFIXES
# STEMMING CAN BE 3 TYPES:
# 1. PORTER STEMMER,
# 2. LANCASTER STEMMER
# 3. SNOWBALL STEMMER
```

```python
## 1. PORTER STEMMER
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
text = ["I bought a fiction.","These sneakers are comfortable","I purchased a cotton shirt!", "We visited place."]
[sentence for sentence in text]
token = [word_tokenize(sentence) for sentence in text]
#print("TOKEN: ",token)

for sentence in token:
  for word in sentence:
    print(stemmer.stem(word), end = " ")

```

```python
# 2. LANCASTER STEMMER
from nltk.stem import LancasterStemmer
from nltk.tokenize import word_tokenize
text = "running, reading better ate."
token = word_tokenize(text)
print(token)
lancaster_stemmer = LancasterStemmer()
result = [lancaster_stemmer.stem(word) for word in token]
print(result)
print(lancaster_stemmer.stem(word) for word in token) # OUTPUT : <generator object <genexpr> at 0x7b0dfead31d0>
print(list(lancaster_stemmer.stem(word) for word in token)) # OUTPUT : ['run', ',', 'read', 'bet', 'at', '.']
```

```python
# 3. SNOWBALL STEMMER - Aggressive and overstemming low and support multiple language
from nltk.stem import SnowballStemmer
from nltk.tokenize import word_tokenize
text = "happiness running got ate in on the better"
token = word_tokenize(text)
stemmer = SnowballStemmer("english")
stemmed_word = [stemmer.stem(w) for w in token]
print(stemmed_word)
```

```python
# LEMMATIZATION -REDUCE THE WORD TO THEIR DICTIONARY FORM (LEMMA)
# Ensure valid words and more accurate than Stemming
# Process : Tokenization -> POS Tagging -> Lemmatization
# Types of Lemmatization :
# 1. RULE BASED (remove - ing, ed, s)
# 2.DICTIONARY BASED (WordNetLemmatizer)
# 3. STATISTICAL (Machine learning model - more accurate)

# Rule-Based Lemmatization → Custom rules (or spaCy's rule tables in some pipelines)
# Dictionary-Based Lemmatization → NLTK WordNetLemmatizer
# Statistical Lemmatization → spaCy en_core_web_sm
```

```python
# 1. Statistical Lemmatization- SpaCy
import spacy
nlp = spacy.load("en_core_web_sm")
text = "happiness running got ate in on the better"
token = word_tokenize(text)
for word in token:
  doc = nlp(word)
  for w in doc:
    print(w.text, w.pos_, w.lemma_)
```

```python
# 2. DICTONARY BASED - WordNetLemmatizer
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
import nltk
nltk.download('wordnet')
text = "happiness running got ate in on the better"
token = word_tokenize(text)
lemmatizer = WordNetLemmatizer()
print(list(lemmatizer.lemmatize(word, pos='a') for word in token)) # By default the Lemmatizer assume it is a noun >>> OUTPUT : better -> good
print(list(lemmatizer.lemmatize(word, pos='v') for word in token)) # By default the Lemmatizer assume it is a noun >>> OUTPUT : running -> run
```

```python
# STOPWORDS REMOVAL - CORPUS
#is, am,are, was,were,do, does,doing, the a an and of in on
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
text = "I am doing the wonderful things. All are good."
token = word_tokenize(text)
stop_words = set(stopwords.words("english"))
print("removing the word:")
for word in token:
  if word.lower() in stop_words:
    print(word, end = ", ")

```

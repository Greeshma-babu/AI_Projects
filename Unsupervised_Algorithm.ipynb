# KMean Algorithm

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

iris = load_iris()
x = iris.data[:,:2]
scaler = StandardScaler()
x_scaler = scaler.fit_transform(x)
print(x_scaler)
```

```python
KMeans = KMeans(n_clusters=3, random_state=4)
KMeans.fit(x_scaler)
```

```python
labels = KMeans.labels_
print(labels)
```

```python
centroid = KMeans.cluster_centers_
print(centroid)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_scaler[:,0], x_scaler[:,1], c=labels, cmap='rainbow')
plt.scatter(centroid[:,0], centroid[:,1], c='black', marker = 'x')
plt.title("KMeans Algorithm")
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.legend()
plt.show()
```

# Hierarchical Clustering

```python
import scipy.cluster.hierarchy as sch
plt.figure(figsize = (10,10))
dendrogram  = sch.dendrogram(sch.linkage(x_scaler, method='ward'), above_threshold_color='green', color_threshold=100)
plt.title('Dendrogram')
plt.xlabel('Data Point')
plt.ylabel('Euclidean Distance')
plt.show()
```

# DBSCAN

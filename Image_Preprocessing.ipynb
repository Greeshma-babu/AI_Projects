```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
for filename in uploaded.keys():
  # cv2.imread() reads the image from disk and returns the image as a NumPy array.
  img  = cv2.imread(filename)
  print("image:", img)

  # Convert BGR to RGB
  # OpenCV loads images in BGR order, while Matplotlib expects RGB order.
  img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
  print("img_rgb",img_rgb)

  plt.imshow(img_rgb)
  plt.show()
```

**IMAGE SCALING**

```python
# scr - Source of the image
# dsize= None new image size (width, height) in pixels,
# Scaling factor along the width and height (x-axis, y- axis).
scaled = cv2.resize(img_rgb, None, fx =0.5, fy =0.8)
plt.imshow(scaled)
plt.show()
```

```python
scaled = cv2.resize(img_rgb, (1000,500))
plt.imshow(scaled)
plt.show()
```

```python
gray = cv2.cvtColor(img_rgb,cv2.COLOR_BGR2GRAY)
# imshow - prepares the image to be drawn but doesn't necessarily display it immediately.
plt.imshow(gray)
# show- Render and display everything that has been prepared.
plt.show()
```

```python
hsv = cv2.cvtColor(img_rgb,cv2.COLOR_BGR2HSV)
plt.imshow(hsv)
plt.show()
```

```python
bright = cv2.convertScaleAbs(img_rgb, alpha=1, beta=100)
plt.imshow(bright)
plt.show()
```

```python
contrast = cv2.convertScaleAbs(img_rgb, alpha=2.5, beta=0)
plt.imshow(contrast)
plt.show()
```

```python
# kernel size - It tells OpenCV how many neighboring pixels to consider when blurring each pixel.
# 3rd parameter-  how much weight is given to neighboring pixels. (small value - less blur)
blur =cv2.GaussianBlur(img_rgb, (25,25),10)
plt.imshow(blur)
plt.show()
```

```python
edge =cv2.Canny(img_rgb, 300,100)
plt.imshow(edge)
plt.show()
```

```python
h,w = img_rgb.shape[:2]
m = cv2.getRotationMatrix2D((w//2,h//2),45,1)
rotated = cv2.warpAffine(img_rgb, m,(w,h))
plt.imshow(rotated)
plt.show()
```

```python
h,w = img_rgb.shape[:2]
m = cv2.getRotationMatrix2D((w//2,h//2),45,5)
rotated = cv2.warpAffine(img_rgb, m,(w,h))
plt.imshow(rotated)
plt.show()
```

```python
flip_h = cv2.flip(img_rgb, -10)
plt.imshow(flip_h)
plt.show()
```

```python
flip_h = cv2.flip(img_rgb, 10)
plt.imshow(flip_h)
plt.show()
```

```python
m = np.float32([[10,0, 50], [0,1,30]])
translated = cv2.warpAffine(img_rgb, m, (img_rgb.shape[1], img_rgb.shape[0]))
plt.imshow(translated)
plt.show()
```

```python
plts1 = np.float32([[10, 50], [20,30],[50,34]])
plts2 = np.float32([[5, 55], [10,30],[50,34]])
translated = cv2.getAffineTransform(plts1, plts2)
sheared = cv2.warpAffine(img, translated, (img.shape[1], img.shape[0]))
plt.imshow(sheared)
plt.show()
```
